```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def nearestNeighbour(newVector,data,labels):
    closestNId = np.argmin(np.sum(np.subtract(data,newVector)**2, axis=1))
    return labels[closestNId]
```

```python
def knearestNeighbours(newVector, data, labels, k = 3):
    KclosestNId = np.argpartition(np.sum(np.subtract(data,newVector)**2, axis=1), k)[:k]
    kLabels = labels[KclosestNId]
    unique, counts = np.unique(kLabels, return_counts=True)
    return unique[np.argmax(counts)]
```
