```python
def generateAllPosibleStartingPoints(numPoints: int, k: int):
    return np.array([i for i in range(2**numPoints) if i.bit_count() == k])
```

```python
startingPoints = generateAllPosibleStartingPoints(10,3)
print(len(startingPoints))
print(startingPoints)
```

```python
def kMeansCluster(nums, k):
    bitmaps = generateAllPosibleStartingPoints(nums, k)
    bool_masks = (bitmaps[:, None] & (1 << np.arange(len(nums)))) > 0
```
